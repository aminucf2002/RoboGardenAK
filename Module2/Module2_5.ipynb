```python
## 2-7 Data visualization Cancer
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler

df=pd.read_csv('data.csv')

#drop missing value
df.dropna(axis=1,how='all',inplace=True)
print('number of nans in each column: \n',df.isna().sum())


features=df[['area_se','smoothness_se','concavity_se','texture_worst','perimeter_worst','area_worst']].copy()
diagnosis=df['diagnosis'].copy()

# Standardize features
scaler=StandardScaler()
for feature in features:
    scaler.fit(features[feature].values.reshape(-1,1))
    features[feature]=scaler.transform(features[feature].values.reshape(-1,1)).reshape(-1)
plt.scatter(features['area_se'],features['smoothness_se'],c=df['diagnosis'].map({'M':1,'B':0}))
plt.xlabel('area_se')
plt.ylabel('smoothness_se')


# Correlation matrix
import seaborn as sns
corr=features.corr()
plt.figure(figsize=(10,8))
sns.heatmap(corr,annot=True,cmap='coolwarm')

```

```python
df=pd.read_excel('Bank_Personal_Loan_Modelling.xlsx',sheet_name='Data')
print(df.head())
#Data Cleaning
#Step 1: Find the Missing Values in the attributes
df.isna().sum()
#Step 2: Missing Values in Mortgage and Family
#For these attributes, we can safely assume that if the data is missing it can be accounted for as zero. 
cols=['Family','Mortgage']
for col in cols:
    df[col]=df[col].fillna(0)
print(df[col].isna().sum())

# Step 3: Missing Values in Age and Experience
# We will fill the values missing with the mean value
cols=['Age','Experience']
for col in cols:
    mean_value=df[col].mean()
    df[col]=df[col].fillna(mean_value)
print(df[col].isna().sum())

# Dealing with Duplicate Values
# Step 1: Find the Duplicate Values in the attributes
df.duplicated(subset=['ID']).sum()

# Step 2: Remove Duplicate Values
df.drop_duplicates(subset=['ID'],inplace=True)
df.duplicated(subset=['ID']).sum()


# Data Preparation
# Step 1: Drop Unneeded columns
df=df.drop(columns=['ID'],axis=1)

# Step 2: View the Zip code data
df_county=pd.read_csv('Zip_County.csv')
df_county.head()


```

```python
# Step 3: Merge the County name to the original dataframe
df=pd.merge(df,df_county,how='inner',on='ZIP Code')
df.head()
##
# Remove Negative values in Experience column
print("Shape Before Filtration:",df.shape)
df=df[df['Experience']>=0]
print("Shape After Filtration:",df.shape)
```
