```python
# RNN
from tensorflow import keras
import numpy as np

from tensorflow.keras.datasets import imdb

(x_train, y_train), (x_test, y_test) = imdb.load_data(
    path="imdb.npz",
    num_words=None,
    skip_top=0,
    maxlen=None,
    seed=113,
    start_char=1,
    oov_char=2,
    index_from=3,
)
#only include the first 1000 data
x_train=x_train[0:120]
y_train=y_train[0:120]
x_test=x_test[0:100]
y_test=y_test[0:100]


```

```python
x_train=np.array(x_train)
print ("size of x_train",x_train.shape)
print("type of x_train", type(x_train))
print("x_train[0]",x_train[0:2])
print("y_train[0]",y_train[0])
```

```python
# Retrieve the word index file mapping words to indices
tokens = keras.datasets.imdb.get_word_index()

#get the size of tokens
len(tokens)
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences

#Use a loop to find the longest subsequence in the dataset
sequence_length = max([len(sequence) for sequence in x_train])
sequence_length =max(sequence_length,50)
# Pad all the subsequences with 0's, and add to the front of the data
sequences = pad_sequences(x_train, maxlen=sequence_length, padding='post')

# Print the shape of the padded sequences
print(sequences.shape)
```

```python
subsequences = []
for sequence in x_train:
    for i in range(1, len(sequence)):
        subsequence = sequence[:i+1]
        subsequences.append(subsequence)
#visualize the new set-up
print(subsequences[1:10])
print(sequences.size)
```

## Add a padding layer around the sequences to standardize the length of our training phrases. 

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences

#Use a loop to find the longest subsequence in the dataset
sequence_length = max([len(sequence) for sequence in sequences])

# Pad all the sequences with 0's, and add to the front of the data
padded_sequences = pad_sequences(sequences, maxlen=sequence_length, padding='post')

# Print the shape of the padded sequences
print("shoape of padded_sequences=",padded_sequences.shape)
#print a sample of sequences
print(padded_sequences[1])
#limit the sequence length to 150
limitedsize = min(sequence_length, 150)
padded_sequences_limitedsize=padded_sequences[:,0:limitedsize]
print("size of padded_sequences_limitedsize=",padded_sequences_limitedsize.shape)
print("size of y-train=", y_train.shape)
```

```python
#train a ANN to preidct y_train from sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dropout, Dense
x_train_encoded=padded_sequences_limitedsize
model = Sequential()
model.add(Dense(units = x_train_encoded.shape[1], activation = 'relu', input_dim = x_train_encoded.shape[1]))
model.add(Dropout(rate = 0.1))
model.add(Dense(units = 50, activation = 'relu'))
model.add(Dropout(rate = 0.1))
model.add(Dense(units = 1, activation = 'sigmoid'))
model.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
model.fit(x_train_encoded, y_train, epochs = 10, batch_size = 32)

#predict y_test from sequences
```

```python
from keras.utils import to_categorical
vocabulary_size = len(tokens) + 1
vocabulary_size
y = to_categorical(y, num_classes=vocabulary_size) # num_classes does (value - 1) columns
```

## 7. Create the RNN to predict the next word

Define an RNN with the following layers:

An embedding layer with the following parameters:
<br> The input dimension is vocabulary_size
<br> The output dimension is 100
<br> The input length is sequence_length - 1
<br> An LSTM layer with 100 units
<br> A dropout layer with a dropout rate of 10%
<br> A dense layer with the following parameters:
<br> The activation function is softmax
<br> The number of units is vocabulary_size

```python
from keras.models import Sequential 
from keras.layers import LSTM 
from keras.layers import Embedding

model = Sequential()
model.add(Embedding(input_dim = vocabulary_size, 
                    output_dim = 100,
                    input_shape = (sequence_length - 1,)))


model.add(LSTM(units = 10))
from keras.layers import Dropout, Dense 
model.add(Dropout(rate=0.1))
model.add(Dense(units=vocabulary_size, activation ='softmax'))
model.summary()
```
