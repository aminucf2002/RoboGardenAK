```python
import pathlib
import tensorflow.keras as keras
print("Downloading dataset...")

url = "http://storage.googleapis.com/download.tensorflow.org/data/fra-eng.zip"

 
# 'extract=True' should unzip it, but sometimes cache prevents re-extraction.
path = keras.utils.get_file("fra-eng.zip", origin=url, extract=True)
download_dir = pathlib.Path(path).parent

# Robustly find 'fra.txt' anywhere in the download directory
text_file_candidates = list(download_dir.glob("**/fra.txt"))


if not text_file_candidates:
    # If not found, force unzip manually
    print("File not found after auto-extract. Unzipping manually...")

    import zipfile
    with zipfile.ZipFile(path, 'r') as zip_ref:
        zip_ref.extractall(download_dir)
    text_file_candidates = list(download_dir.glob("**/fra.txt"))

 

if not text_file_candidates:
    raise FileNotFoundError(f"Could not find 'fra.txt' in {download_dir}")

text_file = text_file_candidates[0]
print(f"Dataset found at: {text_file}")

# Read the file
with open(text_file) as f:
    lines = f.read().split("\n")[:-1]
# Parse pairs (English, French)
text_pairs = []

for line in lines:
    parts = line.split("\t")
    # Some lines might be malformed, so we check length
    if len(parts) >= 2:
        english, french = parts[0], parts[1]
        french = "[start] " + french + " [end]"
        text_pairs.append((english, french))

 

print(f"Successfully loaded {len(text_pairs)} sentence pairs.")
```
