## Use a Weighted Rating
<font color = "blue"> Create a function named popularity Recommender and use it to recommend books based on popularity. <br>
Use a weighted rank similar to that used in the IMDB rating example in lesson 2.
    
![alt text](IMDB.jpg "IMDB Calculation")

```python
df.columns
```

```python
def popularityRecommender(df):
    
    #Define the minimum vote count
    minimum_vote_count = 0.75 * df['vote_count'].max()
    
    #Define C – the mean rating
    mean_rating = df['vote_average'].mean()

    df['weighted_rating'] = (((df['vote_count']/(df['vote_count']+minimum_vote_count)) * df['vote_average']) +
                             ((minimum_vote_count/(df['vote_count']+minimum_vote_count))*mean_rating))

    recommendations = df.sort_values(by = 'weighted_rating',ascending = False).head(5)
    
    return(recommendations)   
```

```python
top5 = popularityRecommender(df)
top5[["title",'vote_count','vote_average','weighted_rating',]].head(5)
```

```python
top5Names = top5["title"].head(5).values
print(top5Names)
```
