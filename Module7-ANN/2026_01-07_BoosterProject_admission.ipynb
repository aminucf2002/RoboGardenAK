## Miniproject 3: Predict Studetn Admission
Load the data from the CSV file.
Drop the first column and split the columns into input and output.
Split the data into training and test sets.
Scale input features using a standard scaler.

```python
# Load the data from the CSV file.
import pandas as pd
from pandas import read_csv
dataset=pd.read_csv('Admission_Predict_Ver1.1.csv')
```

```python

#check for missing data
dataset.isnull().sum()

# Drop the first column and split the columns into input and output.
X = dataset.iloc[:, 1:-1].values
y = dataset.iloc[:, -1].values
# Split the data into training and test sets.
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# Scale input features using a standard scaler.
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)

```

## Step 5: Build and Train the Selected Models

### For each of the selected models:
Use the scikit-learn library to build the model using default parameters.
Use the training set to train the model.


```python
# Import the required libraries
import numpy as np
import pandas as pd 
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense


# Initialize the model using default parameters
model = Sequential()
#creating the input layers
model.add(Dense(7, input_dim=7, activation='relu'))
model.add(Dense(32, activation='relu'))
model.add(Dense(16, activation='relu'))
model.add(Dense(8, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
model.summary()
# Train the model using the training set
model.fit(X_train, y_train, epochs=100, batch_size=32)

# Evaluate the model using the test set
model.evaluate(X_test, y_test)

# Make predictions on new data
y_pred = model.predict(X_test)

# Convert the predictions to binary values
y_pred = (y_pred > 0.5)

# Print the predictions
print(y_pred)

```
