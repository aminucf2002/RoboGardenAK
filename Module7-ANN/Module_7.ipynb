 You are given training and test sets from the Credit-G dataset. Write code that uses the Scikit-Learn library to build, train, and test an ANN classifier with the architecture described below.

 Build the ANN classifier with the following parameters:
The network has an input layer and two hidden layers
 Each layer has 100 units
 Activation function: ReLU
 Optimizer: Adam
 Number of training epochs: 100
Batch size: 20
random_state=0
Train the network using the training data (x_train and y_train).
Calculate the confusion matrix for the model using the test set (x_test and y_test).
 Your code should return the confusion matrix.

```python
def main():
    from sklearn.datasets import fetch_openml
    dataset = fetch_openml(name='credit-g', version=1, as_frame=True)
    x = dataset.data
    y = dataset.target
    from sklearn.preprocessing import LabelEncoder
    encoder = LabelEncoder()
    y = encoder.fit_transform(y)
    from sklearn.model_selection import train_test_split
    x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 0)
    from sklearn.preprocessing import StandardScaler
    scaler = StandardScaler()
    import pandas as pd
    x_train = pd.get_dummies(x_train)
    x_test = pd.get_dummies(x_test)
    x_train, x_test = x_train.align(x_test, join='left', axis=1, fill_value=0)
    x_train = scaler.fit_transform(x_train)
    x_test = scaler.transform(x_test)
    from keras.models import Sequential
    from keras.layers import Dense
    model = Sequential()
    model.add(Dense(units=100, activation='relu'))
    model.add(Dense(units=100, activation='relu'))
    model.add(Dense(units=1, activation='sigmoid'))
    model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
    model.fit(x_train, y_train, batch_size=20, epochs=100)
    y_pred = model.predict(x_test)
    y_pred = (y_pred > 0.5)
    from sklearn.metrics import confusion_matrix
    confusion_matrix = confusion_matrix(y_test, y_pred)

    # Write your code here. Do not change any other parts of the code
    from sklearn.metrics import accuracy_score
    print('Accuracy score: ' + str(accuracy_score(y_test,y_pred)*100))
    return confusion_matrix
main()
```

```python
# Perform the following task in the provided editor to complete this mission:

# Step 1    

# You are given training and test sets from the Credit-G dataset. Write code that uses the Keras library to build, train, and test an ANN classifier with the architecture described below.

# Your code should perform the following tasks to complete this mission:

# Build the ANN with the following layers:

# The input layer is a fully-connected layer with the following parameters:
# - 100 units
# - Activation function: relu

# Two hidden layers are fully-connected layers with the following parameters:
# - 100 units
# - Activation function: relu

# The output layer is a fully-connected layer with:
# - 1 units
# -  Activation function: softmax
# Build the network using the following parameters:
# Optimizer: adam
# Loss Function: binary_crossentropy
# Metrics: accuracy
# Train the network using the following parameters:
# Number of training epochs: 100
# batch size: 20
# Calculate the confusion matrix for the model using the test set (x_test and y_test).
# Your code should return the confusion matrix.

def main():
    from sklearn.datasets import fetch_openml
    dataset = fetch_openml(name='credit-g', version=1)

    x = dataset.data
    y = dataset.target

    from sklearn.preprocessing import LabelEncoder
    encoder = LabelEncoder()
    y = encoder.fit_transform(y)
    import pandas as pd
    x = pd.get_dummies(x, drop_first=True)
    from sklearn.model_selection import train_test_split
    x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 0)

    from sklearn.preprocessing import StandardScaler
    scaler = StandardScaler()
    x_train = scaler.fit_transform(x_train)
    x_test = scaler.transform(x_test)

    # Write your code here. Do not change any other parts of the code
    from keras import Sequential
    from keras.layers import Dense

    model=Sequential()
    model.add(Dense(units=100,activation='relu'))
    model.add(Dense(units=100,activation='relu'))
    model.add(Dense(units=100,activation='relu'))
    model.add(Dense(units=1,activation='softmax'))

    model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
    model.fit(x_train,y_train, batch_size = 20,epochs=100)
    y_pred=model.predict(x_test)
    y_pred = (y_pred > 0.5)

    from sklearn.metrics import confusion_matrix
    confusion_matrix = confusion_matrix(y_test, y_pred)

    from sklearn.metrics import r2_score
    print('R2 squared score: ' + str(r2_score(y_test,y_pred)*100))

    return confusion_matrix
main()
```

# Step 4: Clustering
Use K-Means and Mean-shift to fit your data.
Get optimal k for K-Means using the elbow curve method.
Try different bandwidth for the mean-shift clustering.

```python
#apply k-means and mean-shift
from sklearn.cluster import KMeans, MeanShift, estimate_bandwidth

# Apply K-Means
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans_labels = kmeans.fit_predict(dataset)

# Apply Mean-Shift
bandwidth = estimate_bandwidth(dataset, quantile=0.2, n_samples=500)
meanshift = MeanShift(bandwidth=bandwidth, bin_seeding=True)
meanshift_labels = meanshift.fit_predict(dataset)

# results
print("K-Means Labels:", kmeans_labels)
print("Mean-Shift Labels:", meanshift_labels)

# Visualize results
plt.scatter(dataset.iloc[:, 0], dataset.iloc[:, 1], c=kmeans_labels, cmap='viridis')
plt.title('K-Means Clustering')
plt.show()

plt.scatter(dataset.iloc[:, 0], dataset.iloc[:, 1], c=meanshift_labels, cmap='viridis')
plt.title('Mean-Shift Clustering')
plt.show()
```
