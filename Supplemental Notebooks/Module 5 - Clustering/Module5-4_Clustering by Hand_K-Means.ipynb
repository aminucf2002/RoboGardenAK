In this notebook, we take a look at a simplified version of the code behind a K-Means Clustering Method. 

1) Create a class for the clustering model. This contains a way to take in the data and the number of clusters as arguments. The class contains methods: <br>a) Initialize parameters (__init__), <br>b) Define the clusters (**fit**) which uses the private classes **_assign_clusters** and **_compute_centroids** to assign the closest points to a cluster and then calculate a new centroid. <br>c) **Predict** which uses the converged/final/ centroids to assign class labels to new data.
2) Create / read in some data for clustering.
3) Use the trained model to cluster the data! 

### Create the Clustering (K-Means) Model

```python
import numpy as np

class MyKMeans:
    def __init__(self, n_clusters=3, max_iter=100, tol=1e-4):
        self.n_clusters = n_clusters
        self.max_iter = max_iter
        self.tol = tol
        self.centroids = None
        self.labels = None

    def fit(self, X):
        """ Randomly initialize centroids"""
        np.random.seed(42)
        random_idx = np.random.permutation(X.shape[0])[:self.n_clusters]
        self.centroids = X[random_idx]
        

        for i in range(self.max_iter):
            """ Assign clusters based on the nearest centroid"""
            self.labels = self._assign_clusters(X)
            #print("Each Iteration Centroids:",self.centroids)

            """ Calculate new centroids """
            new_centroids = self._compute_centroids(X)

            # Check for convergence
            ''' check to see how far the new centroid has 'moved', if it's less than the tolerance, we're done!'''
            if np.all(np.linalg.norm(new_centroids - self.centroids, axis=1) <= self.tol):
                break

            self.centroids = new_centroids

    def _assign_clusters(self, X):
        """ Calculate distances from each point to each centroid"""
        distances = np.linalg.norm(X[:, np.newaxis] - self.centroids, axis=2)
        #print("Distances:", distances)
        # Assign the nearest centroid
        #print("Return value:", np.argmin(distances, axis=1))
        return np.argmin(distances, axis=1)

    def _compute_centroids(self, X):
        """ Calculate the centroid for the new cluster centre after the first iteration """
        centroids = np.zeros((self.n_clusters, X.shape[1]))
        for k in range(self.n_clusters):
            points_in_cluster = X[self.labels == k]
            if len(points_in_cluster) > 0:
                centroids[k] = points_in_cluster.mean(axis=0)
        return centroids

    def predict(self, X):
        ''' Use the model to cluster data '''
        if self.centroids is None:
            raise Exception("Model not yet trained. Call fit() first.")
        return self._assign_clusters(X)
```

### Use the Model!

```python
''' Make some sample data '''
X = np.array([
        [1.0, 2.0],
        [1.5, 1.8],
        [5.0, 8.0],
        [8.0, 8.0],
        [1.0, 0.6],
        [9.0, 11.0]
    ])
```

```python
''' Use the model to assign clusters: '''
kmeans = MyKMeans(n_clusters=2)
kmeans.fit(X)

print("Centroids:", kmeans.centroids)
print("Labels:", kmeans.labels)
```

```python
kmeans.predict(X)
```
