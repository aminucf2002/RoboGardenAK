## Create a Memory-Based Collaborative Filtering Recommender

````python
# Preprocessing
ratings = ratings.drop('timestamp', axis = 1)
````

````python
ratings
````

### Create the User-Item Matrix

````python
user_item_m = ratings.pivot(index = 'userId', columns = "movieId", values = 'rating').fillna(0)
user_item_m
````

### Define similarity matrix

````python
from sklearn.metrics.pairwise import cosine_similarity
X_user = cosine_similarity(user_item_m)
X_item = cosine_similarity(user_item_m.T)
````

### Define a function to determine and return suitable recommendations to the user for items they have not interacted with yet.

````python
def movie_recommender(user_item_m, X_user, user, k=20, top_n=10): 
    # Get location of the actual user in the User-Items matrix 
    user_ix = user_item_m.index.get_loc(user)
    print(f"Let's look at user with ID # {user} in row {user_ix}.")

    # Use it to index the User similarity matrix 
    user_similarities = X_user[user_ix]
    print(f"Here is their array of similarities:", user_similarities)

    # obtain the indices of the top k most similar users 
    most_similar_users = user_item_m.index[user_similarities.argpartition(-k)[-k:]] 
    print(f"From this list, the {k} most similar users are:",most_similar_users)
    
    # Obtain the mean ratings of those users for all movies 
    rec_movies = user_item_m.loc[most_similar_users].mean(0).sort_values(ascending=False) 
    print(f"\n\nThese {k} users gave the following average scores, listed highest to lowest, for all the movies in the database:", rec_movies[0:20])
    
    # Discard already seen movies 
    m_seen_movies = user_item_m.loc[user].gt(0)
    print('\n',np.count_nonzero(m_seen_movies), "movies have been seen in this database.")
    
    seen_movies = m_seen_movies.index[m_seen_movies].tolist()
    print("\nThe indices of the seen movies:", seen_movies)
    
    rec_movies = rec_movies.drop(seen_movies).head(top_n)
    print(f"\nThese are the top {top_n} highest average movies from the similar users that have not been seen:",rec_movies )
    
    # return recommendations - top similar users rated movies (and link to a list of movie names by id if you have that info)
    return rec_movies.index.to_frame().reset_index(drop=True)#.merge(movies)
````

**Code Explanation Bit by Bit**

```python
user_item_m = ratings.pivot('userId', "movieId", 'rating').fillna(0)
```
* **`user_item_m = ratings.pivot('userId', "movieId", 'rating')`**: This is a key step in preparing the data for a memory-based recommender system.
    * **`ratings.pivot(...)`**: The `pivot` function in pandas is used to reshape a DataFrame based on column values.  Here, we're transforming the `ratings` DataFrame.
    * **`'userId'`**: This specifies that the 'userId' column from the `ratings` DataFrame will become the *index* (rows) of the new `user_item_m` DataFrame.  Each unique user ID will represent a row.
    * **`"movieId"`**: This specifies that the "movieId" column from the `ratings` DataFrame will become the *columns* of the new `user_item_m` DataFrame. Each unique movie ID will represent a column.
    * **`'rating'`**: This specifies that the 'rating' column from the `ratings` DataFrame will populate the *values* within the `user_item_m` DataFrame. The values will be the ratings given by each user (row) to each movie (column).
    * **Result**:  The `pivot` operation creates a user-item matrix where each row represents a user, each column represents a movie, and the cell at the intersection of a row and column contains the rating given by that user to that movie. If a user hasn't rated a particular movie, there will be a missing value (`NaN`) in that cell initially.

* **`.fillna(0)`**:  After the `pivot` operation, cells in `user_item_m` that represent user-movie pairs where no rating exists will be filled with `NaN` (Not a Number).  `.fillna(0)` replaces all these `NaN` values with `0`. In the context of recommender systems, a `0` rating often implies that the user has not interacted with the movie, and we treat it as a neutral or non-preference.  This is a common approach, but you might choose other strategies (like using the average rating) depending on the specifics of your data and problem.


```python
X_user = cosine_similarity(user_item_m)
X_item = cosine_similarity(user_item_m.T)
```
* **`X_user = cosine_similarity(user_item_m)`**:
    * **`cosine_similarity(user_item_m)`**: This applies the `cosine_similarity` function to the `user_item_m` matrix.
    * **`user_item_m`**:  When `cosine_similarity` is applied to the user-item matrix directly, it calculates the cosine similarity between *rows* of the matrix. In `user_item_m`, each row represents a user's ratings for all movies. Therefore, this line calculates the cosine similarity between each pair of *users*.
    * **`X_user = ...`**: The result, a matrix of user-user cosine similarities, is stored in the variable `X_user`. `X_user[i][j]` will represent the cosine similarity between user `i` and user `j`.

* **`X_item = cosine_similarity(user_item_m.T)`**:
    * **`user_item_m.T`**: `.T` is the transpose operation in NumPy. It transposes the `user_item_m` matrix, meaning rows become columns and columns become rows. After transposing, each *row* now represents a movie, and the columns are users.
    * **`cosine_similarity(user_item_m.T)`**: Applying `cosine_similarity` to the transposed matrix calculates the cosine similarity between the *rows* of the transposed matrix, which are the movies in the original matrix.  This computes the cosine similarity between each pair of *movies* based on user ratings.
    * **`X_item = ...`**: The result, a matrix of item-item cosine similarities, is stored in the variable `X_item`. `X_item[i][j]` will represent the cosine similarity between movie `i` and movie `j`.

    * **`user`**: This is the ID of the user for whom we want to generate recommendations.
    * **`k=20`**: This is a parameter with a default value of 20. `k` will determine the number of most similar users to consider when making recommendations. It's often referred to as the "neighborhood size".
    * **`top_n=10`**: This is another parameter with a default value of 10. `top_n` will determine the number of top movie recommendations to return.

Since our index is user IDs, this line finds the numerical index (starting from 0) of the row corresponding to the user ID `user` within the `user_item_m` DataFrame.

* **`user_similarities = X_user[user_ix]`**:
    * **`X_user[user_ix]`**:  We use `user_ix` (the numerical index of the target user) to index into the `X_user` (user-user similarity) matrix. `X_user` is a 2D array where `X_user[i]` gives the array of cosine similarities of user `i` with all other users (including themselves). So, `X_user[user_ix]` retrieves a 1D array containing the cosine similarity scores of the target user (at index `user_ix`) with every other user in the system.
    * **`user_similarities = ...`**: This array of similarity scores is stored in the `user_similarities` variable.
    * **`print(f"Here is their array of similarities:", user_similarities)`**: This line prints the array of cosine similarity scores between the target user and all other users.

```python
    # obtain the indices of the top k most similar users
    most_similar_users = user_item_m.index[user_similarities.argpartition(-k)[-k:]]

```
* **`most_similar_users = user_item_m.index[user_similarities.argpartition(-k)[-k:]]`**: This is a crucial step to find the top `k` most similar users.
    * **`user_similarities.argpartition(-k)`**:
        * **`user_similarities`**: This is the 1D array of cosine similarities between the target user and all other users.
        * **`.argpartition(-k)`**:  `argpartition` is a NumPy function that performs a partial sort. `argpartition(-k)` finds the indices that would partition the array such that the elements at and after the k-th largest position are the largest k elements (in no particular order within that partition). It's more efficient than fully sorting the entire array when you only need to find the top k values. In this case, `argpartition(-k)` returns an array of indices.  The *last* `k` indices in this resulting array correspond to the indices of the `k` most similar users (but they are not yet sorted by similarity within these top k).
        * **`[-k:]`**:  This slices the result of `argpartition` to take only the *last* `k` indices. These indices point to the positions in the original `user_similarities` array that correspond to the top `k` most similar users.

    * **`user_item_m.index[...]`**:
        * **`user_item_m.index`**: This is the index of the `user_item_m` DataFrame, which are the user IDs.
        * **`[...]`**:  We use the indices obtained from `user_similarities.argpartition(-k)[-k:]` to index into `user_item_m.index`. This effectively retrieves the *user IDs* of the top `k` most similar users.

    * **`most_similar_users = ...`**: The list of user IDs of the top `k` similar users is stored in `most_similar_users`.
    * **`print(f"From this list, the {k} most similar users are:",most_similar_users)`**: This line prints the user IDs of the `k` most similar users identified.

```python

```
* **`rec_movies = user_item_m.loc[most_similar_users].mean(0).sort_values(ascending=False)`**: This calculates the average ratings of the similar users for all movies and sorts them in descending order.
    * **`user_item_m.loc[most_similar_users]`**:
        * **`user_item_m`**: The user-item matrix.
        * **`.loc[most_similar_users]`**:  `.loc` is used for label-based indexing in pandas. `most_similar_users` is a list of user IDs. This selects the rows from `user_item_m` corresponding to the `most_similar_users`.  The result is a DataFrame containing the ratings of only the `k` most similar users for all movies.
    * **`.mean(0)`**:
        * **`.mean(0)`**:  Applies the `.mean()` function to the DataFrame from the previous step. The `0` argument specifies that the mean should be calculated along the *rows* (axis 0). In this context, for each movie (column), it calculates the average rating given by the `most_similar_users`.  This essentially averages the ratings of the similar users for each movie.
    * **`.sort_values(ascending=False)`**:
        * **`.sort_values(ascending=False)`**: Sorts the resulting `Series` (which contains the average ratings for each movie) in descending order. Movies with higher average ratings from similar users will appear at the top.
    * **`rec_movies = ...`**: The sorted Series of average movie ratings is stored in `rec_movies`.
    * **`print(f"\n\nThese {k} users gave the following average scores, listed highest to lowest, for all the movies in the database:", rec_movies[0:20])`**: This prints the top 20 average movie ratings from the similar users (before removing already seen movies).

* **`m_seen_movies = user_item_m.loc[user].gt(0)`**:
    * **`user_item_m.loc[user]`**: Selects the row from `user_item_m` corresponding to the target `user`. This gives a `Series` of ratings given by this user to all movies.
    * **`.gt(0)`**:  `.gt(0)` (greater than 0) performs an element-wise comparison. For each rating in the Series, it checks if it's greater than 0. It returns a boolean `Series` where `True` indicates a rating > 0 (meaning the user has rated the movie, so we consider it as "seen"), and `False` means a rating of 0 (or less, though ratings are typically non-negative in this context).
    * **`m_seen_movies = ...`**: This boolean Series, indicating which movies the user has rated (and thus "seen"), is stored in `m_seen_movies`.
    * **`print('\n',np.count_nonzero(m_seen_movies), "movies have been seen in this database.")`**: This line counts and prints the number of movies the user has rated (i.e., "seen"). `np.count_nonzero(m_seen_movies)` counts the number of `True` values in the `m_seen_movies` boolean Series.

* **`seen_movies = m_seen_movies.index[m_seen_movies].tolist()`**:
    * **`m_seen_movies.index`**: This is the index of the `m_seen_movies` Series, which are the movie IDs.
    * **`m_seen_movies`**: Used as a boolean mask to index `m_seen_movies.index`. This effectively selects only the movie IDs where `m_seen_movies` is `True` (i.e., movies that the user has rated).
    * **`.tolist()`**: Converts the resulting pandas `Index` to a Python list.
    * **`seen_movies = ...`**: The list of movie IDs that the user has already seen is stored in `seen_movies`.

    * **`rec_movies.drop(seen_movies)`**:  `rec_movies` is the Series of average movie ratings from similar users, sorted in descending order. `.drop(seen_movies)` removes the movie IDs present in the `seen_movies` list from `rec_movies`. This ensures that we don't recommend movies that the user has already seen.
    * **`.head(top_n)`**: After dropping the seen movies, `.head(top_n)` selects the top `top_n` movies from the remaining recommendations. These are the top-ranked movies from similar users that the target user has not yet seen.
    * **`rec_movies = ...`**: The final list of top `top_n` recommended movies (as a pandas Series) is stored back into `rec_movies`.
    * **`print(f"\nThese are the top {top_n} highest average movies from the similar users that have not been seen:",rec_movies )`**: This line prints the top `top_n` recommended movies.

    * **`.reset_index(drop=True)`**: Resets the index of the DataFrame to a default integer index (0, 1, 2, ...), and `drop=True` prevents the old index (movie IDs) from becoming a new column in the DataFrame. It just discards the old index and creates a new default one.
    * **`return ...`**: The function returns this DataFrame, which essentially is a list of movie IDs that are recommended.
